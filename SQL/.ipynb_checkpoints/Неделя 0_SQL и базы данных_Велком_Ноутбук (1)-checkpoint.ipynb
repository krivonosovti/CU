# Стартовый ноутбук курса SQL и базы данных

Этот ноутбук — входная точка в курс. Здесь не нужно решать или сдавать задания: через него можно открыть меню курса и перейти к нужным материалам.

## Как пользоваться

1. **Запустите первую ячейку.** Она подключает ноутбук к сервисам курса. Эта первая ячейка запускается всегда во всех семинарах, домашних заданиях и лекциях-воркшопах. Для работы любого ноутбука нужны два файла, они формируются автоматически: `py_workshop.py` скачивается автоматически полностью; `config.yaml` хранит настройки подключений к СУБД он создается автоматически, но нужно заполнить, при запуске первой ячейки появится нужная форма. 
1. Для заполнения `config.yaml`в Welcome ноутбуке вам ничего вводить не нужно, а в Семинаре, ДЗ, Лекции-Воркшопе - будет нужно заполнить ваши данные для postgresql и Greenplum, они подхватываются автоматически по кнопке "Сбросить пароль" **или альтернативный вариант** Для заполнения `config.yaml` зайдите на [Селф-сервис](https://self-service.culab.ru/) и вручную заполните юзера и пароль. 
1. **Запустите вторую ячейку.** Она открывает интерактивное меню с доступными вам курсами и материалами.
1. В меню выберите нужный курс. Оттуда можно перейти к материалам курса, скачать нужный ноутбук и открыть доступные сервисы курса.
1. Если вы перезапустили kernel или заново открыли окружение, просто снова выполните две ячейки по порядку.
1. При любых технических ошибках и проблемах с подключением попробуйте перегрузить kernel (В настройках Kernel -> Restart Kernel) или сервер (File -> Hub Conrol Panel -> Stop My Server). 

> Сам `Welcome.ipynb` — только стартовая точка. Его не нужно сдавать как учебный материал.


```python
import os
import requests
import time
url = "http://sql-exercise.sql-exercise-prod.svc.cluster.local/"
if os.path.exists("py_workshop.py"):
    os.remove("py_workshop.py")
if not os.path.exists("py_workshop.py"):
    response = requests.get(f"{url}get_py_workshop")
if response.status_code == 200:
    with open("py_workshop.py", "wb") as f: f.write(response.content);time.sleep(1)
from py_workshop import *
runner.set_url(url)
runner.logon()
```
