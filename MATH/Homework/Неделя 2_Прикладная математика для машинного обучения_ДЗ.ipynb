Для вектора $\mathbf{x} = (x_1, x_2, \ldots, x_n) \in \mathbb{R}^n$ и числа $p \geq 1$ p-норма (или норма Гёльдера) определяется как

$$
\|\mathbf{x}\|_p = \left( \sum_{i=1}^{n} |x_i|^p \right)^{1/p}.
$$

Частные случаи:

* 1-норма (манхэттенская норма):
$$
\|\mathbf{x}\|_1 = \sum_{i=1}^{n} |x_i|;
$$
    
* 2-норма (евклидова норма):
$$
\|\mathbf{x}\|_2 = \left( \sum_{i=1}^{n} |x_i|^2 \right)^{1/2};
$$
    
* $\infty$-норма (максимальная норма):
$$
\|\mathbf{x}\|_\infty = \max_{1 \leq i \leq n} |x_i|.
$$

В общем случае p — вещественное.

P-метрика задаётся как $\rho_{||\cdot||_{p}}(x, y) = ||x - y||_{p}$ соответственно.
