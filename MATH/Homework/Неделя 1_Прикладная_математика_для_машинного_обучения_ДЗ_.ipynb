## <font color='DarkOrange'> **Задание 1 [0,25 балла]** </font>

Используя библиотеку NumPy, найди матричное произведение $A \cdot B$.

$A = \begin{pmatrix} 12 & 13 & 2 & -8 \\ 42 & -5 & -27 & 30 \\ 54 & 55 & -1 & 20 \\ -31 & -3 & 40 & -5 \end{pmatrix}$, $B = \begin{pmatrix} 63 & -32 & -42 & 4 \\ -6 & 55 & 11 & 1 \\ -1 & 28 & 71 & -22 \\ 35 & -34 & 36 & -15 \end{pmatrix}$.


```python
A = np.array(
    [[12, 13, 2, -8],
    [42, -5, -27, 30],
    [54, 55, -1, 20],
    [ -31, -3, 40, -5]])
B = np.array(
    [[63, -32, -42, 4],
    [-6, 55, 11, 1],
    [-1, 28, 71, -22],
    [ 35, -34, 36, -15]])
A @ B
```

## <font color='DarkOrange'> **Задание 2 [0,25 балла]** </font>

А теперь найди матричное произведение $B \cdot A$, где $A$ и $B$ из прошлой задачи.

```python
B @ A
```

## <font color='DarkOrange'> **Задание 3 [0,5 балла]** </font>

Верно ли, что $A \cdot B = B \cdot A$ (умножение — матричное произведение)?

```python
print(np.array_equal(A @ B, B @ A))
```

## <font color='DarkOrange'> **Задание 4 [0,5 балла]** </font>

Пусть $x \in \mathbb{R}^{5 \times 1},~A \in \mathbb{R}^{5 \times 3}, y \in \mathbb{R}^{3 \times 1}$.

1) <font color='DarkOrange'> **[0,25 балла]** </font> Найди $x^T \cdot A \cdot y$ , где

$x = \begin{pmatrix} 2 \\ 2 \\ 3 \\ 1 \\ -1 \end{pmatrix}$, $A = \begin{pmatrix} 2 & 1 & 3 \\ 9 & -6 & -3 \\ 3 & -2 & -1 \\ 1 & 0 & -9 \\ -1 & 7 & -5 \end{pmatrix}$, $y = \begin{pmatrix} 7 \\ -8 \\ 4 \end{pmatrix}$.

2) <font color='DarkOrange'> **[0,25 балла]** </font> Какой размерности получился ответ?

```python
x = np.array([[2],[2],[3],[1],[-1]])
y = np.array([[7],[-8],[4]])
A = np.array([[2, 1, 3],[9, -6, -3],[3, -2, -1],[1, 0, -9],[-1, 7, -5]])
result = x.T @ A @ y

print(f"Результат: {result}")
print(f"Размерность {result.shape}")
```
