## <font color='DarkOrange'> **Задание 1 [0,75 балла]**</font>

Реши средствами `NumPy` следующую СЛАУ:
$$A \times B \times C \times x = b,$$
где A, B, C — матрицы, x — вектор неизвестных, b — вектор правой части.

А. Через обратную матрицу.

Б. Общим методом через `NumPy`.

Проверь, что найденное решение верное.


```python
A = [[2, 0, 7],
     [4, 3, 8],
     [6, 1, 9]]

B = [[1, 2, 0],
     [0, 1, 1],
     [2, 0, 1]]

C = [[1, 0, 0],
     [0, 1, 0],
     [0, 0, 1]]

b = [4, 5, 6]
```

```python
M = np.array(A, dtype=float) @ np.array(B, dtype=float) @ np.array(C, dtype=float)
b_vec = np.array(b, dtype=float)

x_inverse = np.linalg.inv(M) @ b_vec

x_solve = np.linalg.solve(M, b_vec)

print("ABC =\n", M)
print("Через обратную матрицу:", x_inverse)
print("Через np.linalg.solve:", x_solve)
print("Подстановка:", M @ x_solve)
print("Невязка:", M @ x_solve - b_vec)

assert np.allclose(M @ x_inverse, b_vec)
assert np.allclose(M @ x_solve, b_vec)
assert np.allclose(x_inverse, x_solve)
```

## <font color='DarkOrange'> **Задание 2 [1,5 балла]**</font>

Реализуй функцию для вычисления решения СЛАУ правилами Крамера.

Функция должна проверять применимость правил, напиши тесты для проверки корректность реализации.



```python
def kramer_solve(A: np.ndarray, b: np.ndarray) -> np.ndarray | None:
    A = np.asarray(A, dtype=float)
    b = np.asarray(b, dtype=float)
    if A.ndim != 2 or A.shape[0] != A.shape[1] or A.shape[0] == 0:
        raise ValueError("A должна быть непустой квадратной матрицей")
    if b.ndim != 1 or b.shape[0] != A.shape[0]:
        raise ValueError("b должен быть вектором длины, равной размеру A")
    if not np.isfinite(A).all() or not np.isfinite(b).all():
        raise ValueError("Все элементы должны быть конечными числами")
   
    if np.linalg.matrix_rank(A) < A.shape[0]:
        return None
    
    det_A = np.linalg.det(A)

    if det_A == 0:
        return None

    x = np.empty(len(b))

    for i in range(len(b)):
        A_i = A.copy()
        A_i[:, i] = b
        x[i] = np.linalg.det(A_i) / det_A

    return x
```

```python
for test_A, expected_x in [
    ([[2, 1], [1, -1]], [2, -1]),
    ([[0, 1], [1, 0]], [0, 3]),
    ([[4]], [2.5]),
    (np.eye(3), [0, 0, 0]),
]:
    test_A = np.array(test_A, dtype=float)
    expected_x = np.array(expected_x, dtype=float)
    test_b = test_A @ expected_x
    result = kramer_solve(test_A, test_b)
    assert np.allclose(result, expected_x)
    assert np.allclose(test_A @ result, test_b)

assert kramer_solve([[1, 2], [2, 4]], [3, 6]) is None
assert kramer_solve([[1, 2], [2, 4]], [3, 7]) is None
assert kramer_solve(np.zeros((2, 2)), [0, 0]) is None

# Некорректные входные данные.
for bad_A, bad_b in [
    ([[1, 2, 3], [4, 5, 6]], [1, 2]),
    ([[1, 0], [0, 1]], [1]),
    ([[1, 0], [0, 1]], [[1], [2]]),
    ([], []),
    ([[np.nan]], [1]),
    ([[1]], [np.inf]),
]:
    try:
        kramer_solve(bad_A, bad_b)
    except ValueError:
        pass
    else:
        raise AssertionError("Ожидался ValueError")

rng = np.random.default_rng(42)
for size in range(1, 7):
    for _ in range(10):
        test_A = rng.normal(size=(size, size)) + size * np.eye(size)
        test_b = rng.normal(size=size)
        result = kramer_solve(test_A, test_b)
        assert result is not None
        assert np.allclose(result, np.linalg.solve(test_A, test_b))
        assert np.allclose(test_A @ result, test_b)
print("Все тесты пройдены")
print("Решение системы задания 1 по Крамеру:", kramer_solve(M, b_vec))
```

## <font color='DarkOrange'> **Бонусная задача [2 балла]**</font>

Реши систему матричных уравнений

$$
\begin{cases}
3X_1 + 2X_2 = 31I, \\
-2X_1 + 5X_2 = 34I,
\end{cases}
$$

где $X_1$, $X_2$ и $I$ — матрицы размера $2 \times 2$, а $I$ — единичная матрица.



**Требования**
1. Найди матрицы $X_1$ и $X_2$.
2. Проверь решение подстановкой.
3. Покажи промежуточные вычисления.


<font color='MediumTurquoise'>**Tip:**</font>
сводится к обычным СЛАУ.

```python
I = np.eye(2)
X2 = (164 / 19) * I
X1 = (31 * I - 2 * X2) / 3

print("19X2 =\n", 164 * I)
print("3X1 =\n", 31 * I - 2 * X2)
print("X1 =\n", X1)
print("X2 =\n", X2)
print("3X1 + 2X2 =\n", 3 * X1 + 2 * X2)
print("-2X1 + 5X2 =\n", -2 * X1 + 5 * X2)

assert np.allclose(3 * X1 + 2 * X2, 31 * I)
assert np.allclose(-2 * X1 + 5 * X2, 34 * I)

coefficients = np.array([[3, 2], [-2, 5]], dtype=float)
entry_solutions = np.linalg.solve(coefficients, np.stack([31 * I, 34 * I]).reshape(2, 4))
assert np.allclose(entry_solutions[0].reshape(2, 2), X1)
assert np.allclose(entry_solutions[1].reshape(2, 2), X2)
```
