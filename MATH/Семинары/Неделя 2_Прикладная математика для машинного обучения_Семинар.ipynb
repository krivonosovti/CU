Рассмотрим систему уравнений вида $Xw=y$, где $X \in \mathbb{R}^{n \times m}, w \in \mathbb{R}^{m}, y \in \mathbb{R}^{n}$.

Они возникают в модели линейной регрессии, где $n$ вектор-строк $x_i, i \in {1,...,n}$ матрицы $X$ являются признаковым описанием $n$ объектов из train-выборки, а $w$ — параметры (веса), которые нужно вычислить, чтобы получить $n$ значений целевой переменной (таргета) $y$.

Для одного $i$-го объекта таргет находится как $\langle x_i, w \rangle$.

Ответь, почему, чтобы напрямую вычислить значения $w$, нельзя просто домножить обе части равенства на $X^{-1}$.


### Задача 5

Решается задача оптимизации:

$||Xw - y||_2^2 \to \min_{w}$.

Её решением является

$w = (X^TX)^{-1}X^Ty$
