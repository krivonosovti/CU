## Пример 2. Тестирование новой ML-модели

Команда обучила новую ML-модель рекомендаций и хочет заменить ей текущую модель.

Перед запуском команда рассматривает две гипотезы:

- новая модель **не улучшает** качество рекомендаций;
- новая модель **улучшает** качество рекомендаций.

Если команда решает, что улучшение есть, новую модель выкатывают в прод.  
Если доказательств улучшения недостаточно — оставляют старую.

**Задания**

1. Сформулируйте основную гипотезу $H_0$ и альтернативную гипотезу $H_1$.
2. Что будет означать ошибка I рода?
3. Что будет означать ошибка II рода?
4. Какая ошибка произошла в каждой ситуации:

   - новую модель выкатили, но на самом деле она не лучше старой;
   - новую модель не выкатили, хотя на самом деле она была лучше;
   - новую модель выкатили, и она действительно была лучше;
   - новую модель не выкатили, и она действительно не была лучше.

**Решение**



Сформулируем гипотезы:

$$
H_0:\ \text{новая модель не лучше старой}
$$

$$
H_1:\ \text{новая модель лучше старой}
$$

Ошибка **I рода** — отвергнуть верную $H_0$.

В нашем случае: решить, что новая модель лучше, и выкатить её, хотя на самом деле улучшения нет.

Ошибка **II рода** — не отвергнуть ложную $H_0$.

В нашем случае: оставить старую модель, хотя новая на самом деле лучше.

## Матрица ошибок

Когда мы проводим эксперимет, по его результатам мы наблюдаем какие-то изменния в метриках. Как узнать или заявить, что этим результатам можно верить и у нас есть достаточно оснований принять решение, что мы сделали что-то лучше?

Почему бы просто не посмотреть на наблюдаемую разницу и заявить, что это — эффект от нашей новой механики?
<br><br>

Предположим, что наш эксперимент в действительности **не оказакал эффекта на продукт**, но мы этого до эксперимента не знаем и нам надо принять какое-то решение.
<br>Тогда если просто случайно заявить, что эффекта от эксперимента нет, то мы с вероятностью 50% угадаем, что его действительно нет. С другой стороны, мы можем сказать что эффект есть, и тогда мы с вероятностью 50% допустим **ошибку** (ошибка 1-го рода).
<br><br>

Также легко предположить, что эксперимет в действительности **оказал эффект** на наш продукт.<br>
И здесь такая же развилка, если мы принимаем решение случайно. Мы либо можем угадать что эффект есть, либо допустить **ошибку**, заявив что его нет (это ошибка 2-го рода).


Такое взаимоотношение реальности и наших ее оценок, при которых мы можем допустить ошибки или угадать, описывается **матрицей ошибок**

image.png

Когда мы проверяем гипотезы статистическими критериями, всегда наше изначальное предположение в том, что эффекта на самом деле нет (нулевая гипотеза).
И по итогам мы либо не отвергаем ее, говоря, что эффекта нет, либо говорим что эффект есть и можем совершить ошибку 1-го рода.


Прелесть в том, что именно эту ошибку 1-го рода в статистических критериях мы и контролируем, когда изначально выбираем $\alpha$ на уровне 5%.

Т.е. статистические критерии, при соблюдении предпосылок, гарантируют нам, что, если эффекта нет, а мы принимаем решение, что он есть, то мы будем совершать эту ошибку в 5% случаев.

**Вывод:**
Матрица ошибок описывает взаимосвязь реального, истинного и недоступного положения дел о наличии или отсутствии эффектов и наших принятых решений относительного наличия или отсутствия этих эффектов.

При этом мы можем допустить **Ошибку 1-го рода** или **Ошибку 2-го рода**

Идея статистического вывода позволяет контролировать **Ошибку 1-го рода** на уровне $\alpha$
