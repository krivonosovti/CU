# Метод моментов

У распределения есть теоретические моменты:

$$m_1(\theta)=\mathbb E[X],\qquad m_2(\theta)=\mathbb E[X^2],\ldots$$

По выборке считаем их аналоги:

$$\widehat m_1=\overline X,\qquad \widehat m_2=\overline{X^2}=\frac1n\sum_{i=1}^nX_i^2.$$


Если неизвестен **один параметр**, обычно берём одно уравнение:

$$m_1(\theta)=\widehat m_1.$$

Если неизвестны **два параметра**, берём два:

$$m_1(\theta_1,\theta_2)=\widehat m_1,\qquad m_2(\theta_1,\theta_2)=\widehat m_2.$$

Решение системы и есть оценки методом моментов.

Мы знаем, какой должна быть характеристика распределения **теоретически**, и заменяем её тем, что увидели **в выборке**:

$$\text{теоретический момент}=\text{выборочный момент}\quad\Longrightarrow\quad\widehat\theta_{\text{ММ}}.$$

Метод обычно прост, но полученная оценка не обязана быть несмещённой или наиболее точной.



## Свойства оценок, полученных методом моментов

1. Метод моментов **не гарантирует несмещённость**: всё зависит от конкретной задачи.

2. При стандартных условиях ММ-оценки обычно **состоятельны**: при росте $n$ выборочные моменты приближаются к теоретическим, а построенная по ним оценка — к истинному значению параметра.

3. Эффективность в общем случае **не гарантируется**.

## Пример 1. Распределение Бернулли

Пусть $X_i\sim\operatorname{Bernoulli}(p)$. Тогда

$$\mathbb E[X]=p.$$

Приравниваем к выборочному среднему: $p=\overline X$. Поэтому

$$\boxed{\widehat p_{\text{ММ}}=\overline X=\frac1n\sum_{i=1}^nX_i}.$$


## Пример 2. Распределение Пуассона

Пусть $X_i\sim\operatorname{Poisson}(\lambda)$. Известно:

$$\mathbb E[X]=\lambda.$$

По методу моментов

$$\boxed{\widehat\lambda_{\text{ММ}}=\overline X}.$$

Например, если за 5 часов получили $4,6,3,8,4$ заказа, то $\widehat\lambda=5$ заказов в час.


## Пример 3. Нормальное распределение

Пусть $X_i\sim\mathcal N(\mu,\sigma^2)$ и неизвестны **два параметра**.

Нужны два момента:

$$\mathbb E[X]=\mu,\qquad \mathbb E[X^2]=\mu^2+\sigma^2.$$

Приравниваем их к выборочным:

$$\mu=\overline X,\qquad \mu^2+\sigma^2=\overline{X^2}.$$

Отсюда

$$\boxed{\widehat\mu_{\text{ММ}}=\overline X},\qquad \boxed{\widehat\sigma^2_{\text{ММ}}=\overline{X^2}-\overline X^{\,2}=\frac1n\sum_{i=1}^n(X_i-\overline X)^2}.$$



# Метод максимального правдоподобия

Пусть есть выборка $x_1,\ldots,x_n$ из распределения с неизвестным параметром $\theta$.

Идея ММП: выбрать такое значение $\theta$, при котором **именно наблюдаемая выборка выглядит наиболее правдоподобной**.


Для одного наблюдения имеем функцию

$$f(x_i;\theta).$$

Для независимых наблюдений перемножаем такие выражения:

$$L(\theta)=f(x_1;\theta)\cdot f(x_2;\theta)\cdots f(x_n;\theta).$$

Функция $L(\theta)$ называется **функцией правдоподобия**.


В функции правдоподобия выборка $x_1,\ldots,x_n$ уже известна. Меняем только параметр $\theta$ и смотрим, при каком значении $L(\theta)$ будет наибольшей.

То есть задача ММП сводится к обычному поиску максимума функции по параметру.


Удобнее работать не с произведением, а с логарифмом:

$$\ell(\theta)=\log L(\theta).$$

Логарифм превращает произведение в сумму и не меняет положение максимума.

Дальше действуем как обычно:

$$\ell'(\theta)=0,$$

решаем это уравнение относительно $\theta$ и получаем ММП-оценку.


**Если параметров несколько**

Например, в нормальном распределении неизвестны $\mu$ и $\sigma^2$.

Тогда строим $\ell(\mu,\sigma^2)$ и по очереди приравниваем к нулю производные по каждому параметру:

$$\frac{\partial \ell}{\partial \mu}=0,\qquad \frac{\partial \ell}{\partial \sigma^2}=0.$$

Получаем две оценки: $\widehat\mu$ и $\widehat\sigma^2$.


## Свойства оценок, полученных методом максимального правдоподобия
1. Метод максимального правдоподобия не гарантирует несмещённость: при конечном размере выборки ММП-оценка может быть смещённой.
2. При стандартных условиях ММП-оценки обычно состоятельны: при росте $n$ оценка сходится к истинному значению параметра.
3. При стандартных условиях ММП-оценки асимптотически эффективны: при больших $n$ их дисперсия приближается к минимально возможной, определяемой границей Крамера–Рао.

## Пример 1. Распределение Бернулли

Сервис показал новую форму заказа $n=200$ пользователям. Заказ оформили $k=46$ человек.

Пусть $Y_1,\ldots,Y_n\sim\operatorname{Bernoulli}(p)$.

1. Выпиши $L(p)$ и $\ell(p)$.
2. Найди ММП-оценку $p$.
3. Сравни её с оценкой метода моментов.


**Решение**

Для $x_i\in\{0,1\}$:

$$\mathbb P_p(X_i=x_i)=p^{x_i}(1-p)^{1-x_i}.$$

Обозначим число успехов $S=\sum_{i=1}^n x_i$. Тогда

$$L(p)=\prod_{i=1}^n p^{x_i}(1-p)^{1-x_i}=p^S(1-p)^{n-S}.$$


Логарифмируем и дифференцируем:

$$\ell(p)=S\log p+(n-S)\log(1-p),$$

$$\ell'(p)=\frac Sp-\frac{n-S}{1-p}=0.$$

Отсюда

$$S(1-p)=p(n-S)\Longrightarrow S=np\Longrightarrow\boxed{\widehat p_{\text{ММП}}=\frac Sn=\overline X}.$$


Для $0<S<n$

$$\ell''(p)=-\frac{S}{p^2}-\frac{n-S}{(1-p)^2}<0,$$

поэтому это максимум. При $S=0$ и $S=n$ максимум достигается на границе, и формула $\widehat p=S/n$ всё равно остаётся верной.


В задаче

$$\widehat p=\frac{46}{200}=0.23.$$

Причём

$$\mathbb E[\widehat p]=p,\qquad \operatorname{Var}(\widehat p)=\frac{p(1-p)}n.$$

Значит, здесь ММП-оценка **несмещённая**, **состоятельная** и совпадает с ММ-оценкой.


## Пример 2. Нормальное распределение

Пусть

$$X_1,\ldots,X_n\sim\mathcal N(\mu,\sigma^2),$$

где оба параметра неизвестны.

1. Найди совместные ММП-оценки $\mu$ и $\sigma^2$.
2. Сравни оценку дисперсии со знаменателем $n$ с несмещённой оценкой со знаменателем $n-1$.


**Решение**

Плотность одного наблюдения:

$$f(x_i;\mu,\sigma^2)=\frac1{\sqrt{2\pi\sigma^2}}\exp\!\left(-\frac{(x_i-\mu)^2}{2\sigma^2}\right).$$

По независимости

$$L(\mu,\sigma^2)=(2\pi\sigma^2)^{-n/2}\exp\!\left(-\frac1{2\sigma^2}\sum_{i=1}^n(x_i-\mu)^2\right).$$


Логарифм правдоподобия:

$$\ell(\mu,\sigma^2)=-\frac n2\log(2\pi)-\frac n2\log\sigma^2-\frac1{2\sigma^2}\sum_{i=1}^n(x_i-\mu)^2.$$

По $\mu$:

$$\frac{\partial\ell}{\partial\mu}=\frac1{\sigma^2}\sum_{i=1}^n(x_i-\mu)=0\Longrightarrow\boxed{\widehat\mu_{\text{ММП}}=\overline X}.$$


Теперь положим $s=\sigma^2$. Тогда

$$\frac{\partial\ell}{\partial s}=-\frac{n}{2s}+\frac1{2s^2}\sum_{i=1}^n(x_i-\mu)^2=0.$$

Умножаем на $2s^2$:

$$-ns+\sum_{i=1}^n(x_i-\mu)^2=0.$$

Подставляем $\widehat\mu=\overline X$:

$$\boxed{\widehat\sigma^2_{\text{ММП}}=\frac1n\sum_{i=1}^n(X_i-\overline X)^2}.$$


**Почему здесь не $n-1$?**

Известно:

$$\mathbb E\!\left[\sum_{i=1}^n(X_i-\overline X)^2\right]=(n-1)\sigma^2.$$

Поэтому

$$\mathbb E[\widehat\sigma^2_{\text{ММП}}]=\frac{n-1}{n}\sigma^2.$$

То есть ММП-оценка дисперсии **смещена вниз**.


Несмещённая оценка имеет вид

$$\boxed{S^2=\frac1{n-1}\sum_{i=1}^n(X_i-\overline X)^2},\qquad \mathbb E[S^2]=\sigma^2.$$

Значит, ММП **не обязана быть несмещённой** даже в очень знакомой модели.

При этом $\widehat\sigma^2_{\text{ММП}}$ состоятельна: множитель $(n-1)/n\to1$.


# Итог

**ММ:** заменяем теоретические моменты выборочными и решаем уравнения относительно параметров.

**ММП:** строим $L(\theta)=\prod f(x_i;\theta)$ и выбираем параметр, максимизирующий правдоподобие.

Свойства нужно проверять отдельно: оценка может быть **несмещённой / смещённой**, **состоятельной**, **эффективной**. ММП при регулярных условиях имеет особенно сильные асимптотические свойства, но это не означает несмещённость при конечном $n$.
