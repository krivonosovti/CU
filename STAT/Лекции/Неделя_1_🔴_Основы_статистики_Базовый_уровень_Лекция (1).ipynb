## Функция выживания

**Функция выживания** показывает вероятность того, что случайная величина $X$ окажется больше заданного значения $x$:

$$
S(x)=\mathbb{P}(X>x)
$$

Она связана с функцией распределения $F(x)=\mathbb{P}(X\leq x)$:

$$
S(x)=1-F(x)
$$

Например, если $X$ — время работы устройства, то $S(100)$ — вероятность того, что устройство проработает дольше $100$ часов.

Функция выживания не возрастает: чем больше $x$, тем меньше вероятность того, что $X>x$.

## Геометрическое распределение

Геометрическое распределение описывает **номер попытки, на которой впервые произошёл успех**.

Если вероятность успеха в каждой независимой попытке равна $p$, то

$$
X \sim \mathcal{Geom}(p)
$$

где возможные значения случайной величины:

$$
X \in {1, 2, 3, \ldots}
$$


Например, если вероятность успеха в каждой попытке равна $0.22$, то

$$
X \sim \mathcal{Geom}(p=0.22)
$$

### Выборка из геометрического распределения

````python
from scipy.stats import geom

p = 0.3
geom_rv = geom(p)

print(geom_rv)
print(type(geom_rv))
````

````python
k_values = np.arange(1, 16)
probabilities = geom_rv.pmf(k_values)

plt.bar(
    k_values,
    probabilities,
    alpha=0.3
)

selected_k = np.arange(1, 4)
selected_probabilities = geom_rv.pmf(selected_k)

plt.bar(
    selected_k,
    selected_probabilities
)

plt.xticks(k_values)
plt.xlabel("k")
plt.ylabel("Вероятность")
plt.title("Функция вероятности геометрического распределения")
plt.grid()
plt.show()
````

````python
p_exactly_3 = geom_rv.pmf(3)

print(
    "Вероятность 3: = ",
    round(p_exactly_3, 4)
)
````

### Функция вероятности геометрического распределения

Вероятность того, что первый успех произойдёт ровно на попытке $k$:

$$
P(X=k)=(1-p)^{k-1}p
$$

Первые $k-1$ попыток должны закончиться неудачей:

$$
(1-p)^{k-1}
$$

а попытка $k$ должна быть успешной:

$$
p
$$

Поэтому

$$
P(X=k)=(1-p)^{k-1}p
$$

Например, вероятность первого успеха на третьей попытке:

$$
P(X=3)=(1-p)^2p
$$

При $p=0.22$:

$$
P(X=3)=0.78^2 \cdot 0.22
$$


### Функция распределения геометрического распределения

Вероятность того, что первый успех произойдёт **не позднее попытки $k$**:

$$
\mathbb{P}(X \leq k)=1-(1-p)^k
$$

Это можно получить через противоположное событие: все первые $k$ попыток оказались неудачными

$$
\mathbb{P}(X>k)=(1-p)^k
$$

следовательно,

$$
\mathbb{P}(X\leq k)=1-\mathbb{P}(X>k)=1-(1-p)^k
$$

**Вероятность успеха не раньше попытки $k$**

Событие $X \geq k$ означает, что первые $k-1$ попыток завершились неудачей:

$$
\mathbb{P}(X\geq k)=(1-p)^{k-1}
$$

При этом

$$
\mathbb{P}(X>k)=(1-p)^k
$$

````python
p = 0.38
geom_rv = geom(p)

p_exactly_3 = geom_rv.cdf(3)

print(
    "Вероятность меньше либо равно 3: = ",
    round(p_exactly_3, 4)
)
````

````python
p_exactly_3 = geom_rv.sf(3)

print(
    "Вероятность больше 3: = ",
    round(p_exactly_3, 4)
)
````

### Свойство отсутствия памяти

Геометрическое распределение обладает **свойством отсутствия памяти**.

Пусть $X$ — номер попытки, на которой произошёл первый успех. Тогда

$$
\mathbb{P}(X>m+n \mid X>m)=\mathbb{P}(X>n)
$$

Это означает: если первые $m$ попыток закончились неудачей, вероятность ждать ещё больше $n$ попыток такая же, как если бы мы начинали испытания заново.

Для геометрического распределения

$$
\mathbb{P}(X>k)=(1-p)^k
$$

По формуле условной вероятности:

$$
\mathbb{P}(X>m+n \mid X>m)=\frac{\mathbb{P}(X>m+n)}{\mathbb{P}(X>m)}
$$

Подставим вероятности:

$$
\frac{(1-p)^{m+n}}{(1-p)^m}=(1-p)^n
$$

Но $(1-p)^n=\mathbb{P}(X>n)$, поэтому

$$
\mathbb{P}(X>m+n \mid X>m)=\mathbb{P}(X>n)
$$

Например, если вероятность успеха равна $p=0.2$, то после десяти неудачных попыток вероятность того, что следующие три попытки тоже будут неудачными, всё равно равна

$$
(1-0.2)^3=0.512
$$

Предыдущие десять неудач на эту вероятность не влияют.

## Логнормальное распределение

Если $X \sim \mathcal{N}(\mu, \sigma^2)$, то $Y = e^X$ имеет логнормальное распределение:
$$Y \sim Lognorm(\mu, \sigma^2)$$

Обратное преобразование:

$$X=\text{ln} Y \sim Norm(\mu, \sigma^2)$$

image.png

### Плотность логнормального распределения

Распишем функцию распределения:

$$F_Y(y) = P(Y \leq y) = P(e^X \leq y) = P(X \leq \ln y) = F_X(\ln y)$$

Дифференцируем функцию распределения для получения функции плотности:

$$f_Y(y) = \frac{d}{dy}F_Y(y) = \frac{d}{dy}F_X(\ln y) = f_X(\ln y) \cdot \frac{d}{dy}(\ln y) = f_X(\ln y) \cdot \frac{1}{y}$$

Подставляем плотность нормального распределения:

$$f_X(\ln y) = \frac{1}{\sigma\sqrt{2\pi}} e^{-\frac{(\ln y - \mu)^2}{2\sigma^2}}$$

Получаем окончательную формулу:

$$f_Y(y) = \frac{1}{y\sigma\sqrt{2\pi}} e^{-\frac{(\ln y - \mu)^2}{2\sigma^2}}, \quad y > 0$$

````python
from scipy.stats import lognorm
````

````python
mu1, sigma1 =  0.5, 0.2
mu2, sigma2 =  1, 0.5
mu3, sigma3 =  0.5, 0.5
````

````python
x = np.linspace(0, 6, 1000)
pdf1 = lognorm.pdf(x, mu1, sigma1)
pdf2 = lognorm.pdf(x, mu2, sigma2)
pdf3 = lognorm.pdf(x, mu3, sigma3)

plt.plot(x, pdf1, label=f'mu={mu1}, sigma={sigma1}')
plt.plot(x, pdf2, label=f'mu={mu2}, sigma={sigma2}')
plt.plot(x, pdf3, label=f'mu={mu3}, sigma={sigma3}')

plt.title('Теоретическая плотность логнормального распределения')
plt.xlabel('x')
plt.ylabel('Плотность вероятности')
plt.legend()
plt.grid()
plt.show()
````

### Свойство отсутствия памяти экспоненциального распределения

Пусть

$$
X \sim \mathrm{Exp}(\lambda), \qquad \lambda > 0
$$

Для любых $s, t \geq 0$ выполняется

$$
\mathbb{P}(X > s + t \mid X > s) = \mathbb{P}(X > t)
$$

По определению условной вероятности

$$
\mathbb{P}(X > s+t \mid X > s)
=
\frac{\mathbb{P}(X > s+t)}{\mathbb{P}(X > s)}
$$

Для экспоненциального распределения

$$
\mathbb{P}(X > x) = e^{-\lambda x}
$$

Следовательно,

$$
\mathbb{P}(X > s+t \mid X > s)
=
\frac{e^{-\lambda(s+t)}}{e^{-\lambda s}}
=
e^{-\lambda t}
=
\mathbb{P}(X > t)
$$

То есть вероятность ждать ещё $t$ единиц времени не зависит от того,
сколько времени $s$ уже прошло.

Поэтому экспоненциальное распределение **обладает свойством отсутствия памяти**.

## Распределение Лапласа

**Распределение Лапласа** — непрерывное распределение, симметричное относительно параметра $\mu$ и имеющее более тяжёлые хвосты, чем нормальное распределение.

Обозначение:

$X \sim \mathcal{Laplace}(\mu, b)$

где $\mu$ — центр распределения, а $b>0$ — параметр масштаба.

Плотность распределения:

$$
f(x)=\frac{1}{2b}\exp\left(-\frac{|x-\mu|}{b}\right)
$$

Распределение симметрично относительно $\mu$

По сравнению с нормальным распределением, распределение Лапласа имеет более острый пик около среднего и более тяжёлые хвосты. Поэтому большие отклонения от среднего для него встречаются чаще.

Например, при $\mu=0$ и $b=1$:

$$
f(x)=\frac{1}{2}e^{-|x|}
$$

Распределение Лапласа часто используют, когда данные сосредоточены около некоторого значения, но при этом редкие большие отклонения встречаются чаще, чем предполагает нормальное распределение.

## Гамма-распределение

**Гамма-распределение** — непрерывное распределение для положительных случайных величин. Его часто используют для моделирования времени ожидания, длительности процессов и положительных величин с правой асимметрией.

Обозначение:

$X \sim \mathcal{Gamma}(\alpha,\theta)$

где $\alpha>0$ — параметр формы, а $\theta>0$ — параметр масштаба.

Плотность распределения:

$$
f(x)=\frac{x^{\alpha-1}e^{-x/\theta}}
{\Gamma(\alpha)\theta^\alpha},
\quad x>0
$$

Здесь $\Gamma(\alpha)$ — гамма-функция.



## Гамма-функция

**Гамма-функция** — это обобщение факториала на положительные вещественные числа.

Она определяется как

$$
\Gamma(\alpha)=\int_0^\infty x^{\alpha-1}e^{-x},dx
$$

Для положительных целых чисел выполняется связь с факториалом:

$$
\Gamma(n)=(n-1)!
$$

Например:

* $\Gamma(1)=1$
* $\Gamma(2)=1$
* $\Gamma(3)=2$
* $\Gamma(4)=6$
* $\Gamma(5)=24$

То есть, например, $\Gamma(5)=4!$.

Важное свойство гамма-функции:

$$
\Gamma(\alpha+1)=\alpha\Gamma(\alpha)
$$

Именно оно похоже на правило для факториала: $(n+1)!=(n+1)n!$.

Главное отличие от обычного факториала — гамма-функцию можно вычислять и для нецелых значений. Например,

$$
\Gamma\left(\frac{1}{2}\right)=\sqrt{\pi}
$$

В гамма-распределении гамма-функция используется как нормирующий множитель:

$$
f(x)=\frac{x^{\alpha-1}e^{-x/\theta}}
{\Gamma(\alpha)\theta^\alpha}
$$

Она нужна для того, чтобы площадь под плотностью распределения была равна $1$.


**Визуализируем фукнцию плотности гамма-распределения для разных значений параметров.**

````python
from scipy.stats import gamma

x = np.linspace(0.01, 20, 500)

params = [
    (1, 1),
    (2, 1),
    (5, 1),
    (2, 4),
    (5, 4)
]

for alpha, theta in params:
    pdf = gamma.pdf(
        x,
        a=alpha,
        scale=theta)

    plt.plot(x,pdf, label=f"α={alpha}, θ={theta}")

plt.xlabel("x")
plt.ylabel("Плотность")
plt.title("Гамма-распределение при разных параметрах")
plt.legend()
plt.grid()
plt.show()
````

### Основные характеристики гамма распределения

- При малых $\alpha$ распределение сильно скошено вправо
- При увеличении $\alpha$ распределение становится более симметричным
- Параметр $\theta$ растягивает или сжимает распределение по горизонтали.

Гамма-распределение удобно использовать для моделирования **времени до наступления нескольких событий**. Например, если события происходят независимо с постоянной интенсивностью, то время ожидания до $\alpha$-го события может иметь гамма-распределение.

При $\alpha=1$ гамма-распределение превращается в экспоненциальное распределение.

# Шпаргалка: распределения из лекции


```python
import numpy as np
from scipy import stats
```

| Название распределения | Его запись | Как вызвать в Python |
| --- | --- | --- |
| **Бернулли** | X ∼ Bern(p) | `stats.bernoulli(p=p)` |
| **Равномерное дискретное** | X ∼ U({a, a+1, …, b}) | `stats.randint(low=a, high=b+1)` |
| **Биномиальное** | X ∼ Bin(n, p) | `stats.binom(n=n, p=p)` |
| **Пуассона** | X ∼ Pois(λ) | `stats.poisson(mu=lam)` |
| **Геометрическое** | X ∼ Geom(p), X ∈ {1, 2, …} | `stats.geom(p=p)` |
| **Равномерное непрерывное** | X ∼ U([a, b]) | `stats.uniform(loc=a, scale=b-a)` |
| **Нормальное** | X ∼ N(μ, σ²) | `stats.norm(loc=mu, scale=sigma)` |
| **Логнормальное** | Y ∼ Lognorm(μ, σ²)<br>ln Y ∼ N(μ, σ²) | `stats.lognorm(s=sigma, scale=np.exp(mu))` |
| **Экспоненциальное** | X ∼ Exp(λ) | `stats.expon(scale=1/lam)` |
| **Лапласа** | X ∼ Laplace(μ, b) | `stats.laplace(loc=mu, scale=b)` |
| **Гамма** | X ∼ Gamma(α, θ) | `stats.gamma(a=alpha, scale=theta)` |

**Обозначения и параметры:**

- `p` — вероятность успеха, `n` — число испытаний.
- `mu` = μ, `sigma` = σ, `lam` = λ, `alpha` = α, `theta` = θ.
- Для дискретного равномерного распределения `a` и `b` — целые границы, обе включены; `high=b+1`, поскольку верхняя граница в `randint` исключается.
- Здесь нормальное распределение записано как N(μ, σ²), где σ² — дисперсия. В `stats.norm(..., scale=sigma)` передаётся стандартное отклонение σ. В лекции встречаются обе записи: N(μ, σ) и N(μ, σ²).
- Для логнормального распределения μ и σ — параметры нормального распределения **логарифма** величины. Правильный вызов: `stats.lognorm(s=sigma, scale=np.exp(mu))`.
- Для непрерывного равномерного распределения `scale=b-a`.
- Для экспоненциального распределения λ — интенсивность, поэтому `scale=1/lam`. Для Пуассона λ — среднее число событий за выбранный интервал.
- Для гамма-распределения α — форма, θ — масштаб. Для распределения Лапласа μ — центр, b — масштаб.

**Как пользоваться созданным распределением:**

```python
rv = stats.norm(loc=0, scale=1)
sample = rv.rvs(size=100)  # сгенерировать выборку
rv.pdf(0)                # плотность в точке (непрерывное распределение)
rv.cdf(1)                # P(X <= 1)
rv.sf(1)                 # P(X > 1)

rv_discrete = stats.binom(n=30, p=0.5)
rv_discrete.pmf(15)      # P(X = 15) для дискретного распределения
```
