```python
from scipy.stats import t  # распределение Стьюдента

x = np.linspace(-5, 5, 1000)

plt.figure(figsize=(9, 5))

for df in [1, 2, 5, 10, 30]:
    y = t.pdf(x, df=df)  # плотность; df — число степеней свободы
    plt.plot(x, y, label=f"df = {df}")

plt.title("Плотность распределения Стьюдента")
plt.xlabel("x")
plt.ylabel("Плотность")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

## Разные и одинаковые дисперсии в t-тесте

Расчёт t-критерия также зависит от предположения относительно **равенства дисперсий в двух выборках**.

 - Если дисперсии **не равны**, то получаем вариант выше, который называем **критерием Уэлча**.

 - Если дисперсии **равны**, то получаем альтернативный вариант расчёта, который называем **критерием Стьюдента**.

**Критерий Стьюдента** (**равные дисперсии**)

$$t = \frac{\bar{X}_1 - \bar{X}_2}{s_p \sqrt{\frac{1}{n_1} + \frac{1}{n_2}}} \sim t_{n_1+n_2-2}$$

$$s_p = \sqrt{\frac{(n_1 - 1)s_1^2 + (n_2 - 1)s_2^2}{n_1 + n_2 - 2}}$$

**Критерий Уэлча** (**разные дисперсии**)

$$t = \frac{\bar{X}_1 - \bar{X}_2}{\sqrt{\frac{s_1^2}{n_1} + \frac{s_2^2}{n_2}}} \sim \nu$$

$$
\nu\approx
\frac{
\left(\frac{s_1^2}{n_1}+\frac{s_2^2}{n_2}\right)^2
}{
\frac{s_1^4}{n_1^2(n_1-1)}
+
\frac{s_2^4}{n_2^2(n_2-1)}
}.
$$

Нетрудно, заметить, что при равных дисперсиях критерий Уэлча сходится к критерию Стьюдента.

Таким образом, можем назвать критерий Стьюдента **частным случаем** критерия Уэлча.

**Поэтому всегда по умолчанию можем пользоваться критерием Уэлча.**
