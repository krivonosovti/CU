# (1 балл) Задача 1

Известно, что рост взрослых мужчин в популяции распределен нормально со средним $175$ см и стандартным отклонением $8$ см. В выборке из $25$ мужчин из одного города средний рост оказался $178$ см.

Проверь гипотезу на уровне значимости $5\%$, что средний рост в этом городе не отличается от общего по популяции.

В этой задаче следует использовать Z-тест, так как дисперсия априори известна.

```python
### Твой код здесь

from scipy.stats import norm

alpha = 0.05
mu0 = 175
sigma = 8
n = 25
mean = 178

z = (mean - mu0) / (sigma / np.sqrt(n))
p_value = 2 * norm.sf(abs(z))

print("H0: средний рост равен 175 см")
print("H1: средний рост отличается от 175 см")
print("Z-статистика:", z)
print("p-value:", p_value)

if p_value < alpha:
    print("Отвергаем H0: средний рост статистически значимо отличается от 175 см.")
else:
    print("Не отвергаем H0: статистически значимое отличие от 175 см не выявлено.")
```

# (1 балл) Задача 2

Для t-распределения с $10$ степенями свободы найдите:

 - $\mathbb{P}(T > 2.5)$

 - Критическое значение для $\alpha = 5\%$ (двусторонний тест)

```python
### Твой код здесь

from scipy.stats import t

df = 10
alpha = 0.05

probability = t.sf(2.5, df)
critical = t.ppf(1 - alpha / 2, df)

print("P(T > 2.5):", probability)
print("Критические значения:", -critical, critical)
```

# (1 балл) Задача 3

Производитель утверждает, что его батарейки работают $100$ часов. Вы тестируете $16$ батареек и получаете данные, которые хранятся в векторе *data*.

Проверь гипотезу, что реальное время работы отличается от заявленного.

Воспользуйся одновыборочным t-критерием.

Попробуй изучить самостоятельно функцию **ttest_1samp** из **scipy.stats** и применить её для ответ на вопрос.

```python
data = [98, 102, 104, 96, 101, 97, 105, 99, 103, 100, 97, 102, 99, 101, 98, 103]

from scipy.stats import ttest_1samp

alpha = 0.05
s, p = ttest_1samp(data, popmean=100, alternative="two-sided")

print("H0: среднее время работы равно 100 часам")
print("H1: среднее время работы отличается от 100 часов")
print("Среднее время работы:", np.mean(data))
print("t-статистика:", s)
print("p-value:", p)

if p < alpha:
    print("Отвергаем H0: среднее время работы статистически значимо отличается от 100 часов.")
else:
    print("Не отвергаем H0: статистически значимое отличие от 100 часов не выявлено.")
```

# (1 балл) Задача 4

Сравни эффективность двух методов обучения.

Проверь гипотезу о равенстве средних между граппами А и В с помощью двухвыборочного t-теста с уровнем значимости $5\%$.

```python
group_a = [85, 88, 90, 82, 87]
group_b = [78, 82, 80, 85, 79]

from scipy.stats import ttest_ind

alpha = 0.05
s, p = ttest_ind(group_a, group_b, equal_var=False, alternative="two-sided")

print("H0: средние в группах A и B равны")
print("H1: средние в группах A и B различаются")
print("Среднее в группе A:", np.mean(group_a))
print("Среднее в группе B:", np.mean(group_b))
print("t-статистика:", s)
print("p-value:", p)

if p < alpha:
    print("Отвергаем H0: средние статистически значимо различаются, среднее в группе A выше.")
else:
    print("Не отвергаем H0: статистически значимое различие средних не выявлено.")
```

# (1 балл) Задача 5

Для F-распределения с df1=5, df2=10 найдите:

 - $\mathbb{P}(F > 3.5)$

 - Критическое значение для $\alpha = 0.05$

```python
### Твой код здесь

from scipy.stats import f

df1 = 5
df2 = 10
alpha = 0.05

probability = f.sf(3.5, df1, df2)
critical = f.ppf(1 - alpha, df1, df2)

print("P(F > 3.5):", probability)
print("Критическое значение:", critical)
```

# (1 балл) Задача 6

Сравни дисперсии двух групп данных на предмет значимых различий с помощью F-теста.

```python
group_a = [23, 25, 28, 22, 26, 24, 27]
group_b = [30, 32, 29, 31, 28, 33, 30]

alpha = 0.05
var_a = np.var(group_a, ddof=1)
var_b = np.var(group_b, ddof=1)

F_calc = var_a / var_b
df1 = len(group_a) - 1
df2 = len(group_b) - 1

lower_crit = f.ppf(alpha / 2, df1, df2)
upper_crit = f.ppf(1 - alpha / 2, df1, df2)
p_value = 2 * min(f.cdf(F_calc, df1, df2), f.sf(F_calc, df1, df2))

print("H0: дисперсии групп A и B равны")
print("H1: дисперсии групп A и B различаются")
print("Дисперсия A:", var_a)
print("Дисперсия B:", var_b)
print("F-статистика:", F_calc)
print("Критические границы:", lower_crit, upper_crit)
print("p-value:", p_value)

if p_value < alpha:
    print("Отвергаем H0: дисперсии статистически значимо различаются.")
else:
    print("Не отвергаем H0: статистически значимое различие дисперсий не выявлено.")
```

# (4 балла) Задача 5. Конверсии

Вы работаете в агрегаторе доставки.

Проводится A/B тест, нацеленный на повешение конверсии в доставку как в ресторанных заказах, так и магазинных.

Есть датасет *delivery_logs.xlsx* с информацией о заказах:

 - *id_order* - уникальный идентификатор заказа;
 - *id_group* - идентфиикатор группы (0 - контроль, 1 - тест);
 - *order_time* - дата+время заказа;
 - *del_time* - дата+время доставки (значение *2000-01-01 00:00:00* означает отсутствие факта доставки);
 - *type* - тип (ресторан или магазин);

**Задача**
  
  1. Сравни конверсию в доставку в магазинах и в ресторанах. Можно ли сказать, что между ними есть статистически значимая разница? (**1 балл**);

  2. Сравни конверсию в доставку в контрольной и в тестовой группе. Можно ли сказать, что между ними есть статистически значимая разница? (**1 балл**);

  3. Сравни конверсию в доставку в контрольной и в тестовой группе только для *ресторанных* заказов. Можно ли сказать, что между ними есть статистически значимая разница? (**1 балл**);

  4. Сравни конверсию в доставку в контрольной и в тестовой группе только для *магазинных* заказов. Можно ли сказать, что между ними есть статистически значимая разница? (**1 балл**);

```python
# dfd = pd.read_excel('resources/week8/delivery_logs.xlsx') 
dfd = pd.read_excel('delivery_logs.xlsx')
dfd['order_time'] = pd.to_datetime(dfd['order_time'], format = '%Y/%m/%d %H:%M:%S')
dfd['del_time']   = pd.to_datetime(dfd['del_time']  , format = '%Y/%m/%d %H:%M:%S')

dfd['flag_delivery'] = np.where(dfd['del_time']=='2000-01-01 00:00:00', 0, 1)

dfd.head()
```

```python
### Твой код здесь

alpha = 0.05

shop = dfd[dfd['type'] == 'Shop']
restaurant = dfd[dfd['type'] == 'Restaurant']

comparisons = [
    ("1. Магазины и рестораны",
     shop['flag_delivery'], restaurant['flag_delivery'], "Магазины", "Рестораны"),
    ("2. Контроль и тест",
     dfd[dfd['id_group'] == 0]['flag_delivery'],
     dfd[dfd['id_group'] == 1]['flag_delivery'], "Контроль", "Тест"),
    ("3. Контроль и тест: рестораны",
     restaurant[restaurant['id_group'] == 0]['flag_delivery'],
     restaurant[restaurant['id_group'] == 1]['flag_delivery'], "Контроль", "Тест"),
    ("4. Контроль и тест: магазины",
     shop[shop['id_group'] == 0]['flag_delivery'],
     shop[shop['id_group'] == 1]['flag_delivery'], "Контроль", "Тест"),
]

for title, sample1, sample2, name1, name2 in comparisons:
    s, p = ttest_ind(sample1, sample2, equal_var=False, alternative="two-sided")

    print(title)
    print("H0: конверсии равны")
    print("H1: конверсии различаются")
    print(f"{name1}: {sample1.mean():.2%}")
    print(f"{name2}: {sample2.mean():.2%}")
    print("t-статистика:", s)
    print("p-value:", p)

    if p < alpha:
        print("Отвергаем H0: конверсии статистически значимо различаются.")
        print("Конверсия выше:", name1 if sample1.mean() > sample2.mean() else name2)
    else:
        print("Не отвергаем H0: статистически значимое различие конверсий не выявлено.")
    print()
```
