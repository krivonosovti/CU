# Задача 1. Геометрическое распределение

Пусть $X$ — номер попытки, на которой впервые произошёл успех. Тогда

$$
\mathbb P(X=k)=p(1-p)^{k-1},\qquad k=1,2,\ldots,
$$

где $p\in(0,1)$ — неизвестная вероятность успеха в одной попытке.

Известно, что

$$
\mathbb E[X]=\frac1p.
$$

По выборке $X_1,\ldots,X_n$:

1. найдите оценку $p$ методом моментов;
2. найдите оценку $p$ методом максимального правдоподобия;
3. вычислите оценку по данным

$$
2,\ 1,\ 4,\ 3,\ 2,\ 1,\ 2,\ 5.
$$


### Пункт 1. Метод моментов
Для геом распределения 
$$
\mathbb E[X]=\frac1p.
$$

$$
\frac1p = \overline X  
$$

Отсюда:
$$
\boxed{\widehat p = \frac{1}{\overline X }}
$$

### Пункт 2. ММП

Плотность:
$$\
f(x; \theta) = \mathbb P(X=k)=p(1-p)^{k-1},\qquad k=1,2,\ldots,
$$

$$
L(p) =  \mathbb \prod_{i=1}^{n}p(1-p)^{X_i-1} = p^n * (1-p)^{\sum(X_i - 1)};
$$

$$
l(p) =  \ln(p^n * (1-p)^{\sum X_i - 1}) = n\ln(p) + (\sum_{i=1}^n (X_i - 1)) * \ln(1-p);
$$

$$
\frac{dl(p)}{dp} = \frac{n}{p} - \frac{\sum_{i=1}^n (X_i - 1)}{1-p}, \qquad l(p)'' < 0;
$$

$$
\frac{n}{p} = \frac{\sum_{i=1}^n (X_i) - n}{1-p};
$$

$$
\boxed{\hat p = \frac{n}{\sum_{i=1}^n (X_i)} = \frac{1}{\overline X}.}
$$



```python
### Пункт 3. Оценка
X_i = np.array([2, 1, 4, 3, 2, 1, 2, 5])

p_hat = 1 / X_i.mean()
print(f"p_hat = {p_hat}")
```

# Задача 2. Распределение Парето

Пусть $X_1,\ldots,X_n$ — независимая выборка из распределения Парето с плотностью

$$
f(x;\alpha)=\alpha x^{-(\alpha+1)},\qquad x\ge1,
$$

где $\alpha>1$ — неизвестный параметр.

Известно, что

$$
\mathbb E[X]=\frac{\alpha}{\alpha-1}.
$$

1. Найдите оценку $\alpha$ методом моментов.
2. Найдите оценку $\alpha$ методом максимального правдоподобия.
3. Объясните, почему две оценки в этой задаче не обязаны совпадать.


### Пункт 1. Метод моментов
$$
\mathbb E[X] = \int_1^{\infty} x \cdot \alpha x^{-(\alpha+1)}\,dx
= \alpha\int_1^{\infty} x^{-\alpha}\,dx
= \left.\frac{\alpha}{1-\alpha}x^{1-\alpha}\right|_1^{\infty}.
$$

При $\alpha>1$ имеем $x^{1-\alpha}\to0$ при $x\to\infty$, поэтому
$$
\mathbb E[X] = \frac{\alpha}{1-\alpha}(0-1) = \frac{\alpha}{\alpha-1}.
$$

$$
\overline X = \frac{\alpha}{\alpha-1}.
$$

$$
\boxed{\hat \alpha = \frac{\overline X}{\overline X - 1}}
$$

### Пункт 2. ММП

$$
f(x;\alpha)=\alpha x^{-(\alpha+1)},\qquad x\ge1,
$$

$$
L(\alpha) = \prod_1^n(\alpha x^{-(\alpha+1)})
$$

$$
l(\alpha) = \sum_1^n\ln(\alpha x^{-(\alpha+1)}) = \sum_1^n\ln(\alpha x^{-(\alpha+1)}) = \sum_1^n\ln(\alpha) -(\alpha+1)\ln(x) = n\ln(\alpha) - (\alpha+1)\sum_{i=1}^n \ln(X_i)
$$

$$
\frac{d}{d\alpha} n\ln(\alpha) - (\alpha+1)\sum_{i=1}^n \ln X_i = \frac{n}{\alpha}  - \sum_{i=1}^{n}\ln{X_i},  \quad l''(\alpha)= - \frac{n}{\alpha^2} < 0
$$

$$
\frac{n}{\alpha}  = \sum_{i=1}^{n}\ln{X_i}
$$

$$
\boxed{\hat \alpha  = \frac{n}{\sum_{i=1}^{n}\ln{X_i}}}
$$

### Пункт 3
Метод моментов использует среднее значений, а ММП среднее логарифмов. Выборки с одинаковым средним могут иметь разные средние логарифмов, поэтому оценка методом моментов будет одинаковой, а оценка ММП — разной. Следовательно, две оценки не обязаны совпадать на конечной выборке.
