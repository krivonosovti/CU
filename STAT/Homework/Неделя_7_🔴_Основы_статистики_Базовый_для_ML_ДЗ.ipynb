# (1 балл) Задача 1. Время восстановления после тренировки

Тренеры фитнес-центра разработали новую программу тренировок и хотят проверить, уменьшает ли она среднее время восстановления после силовой нагрузки.

Известно, что среднее время восстановления равно $\mu_0=60$ минут, а дисперсия времени восстановления равна $\sigma^2=36$.

Для проверки программы тренеры планируют сформировать группу из $n=50$ клиентов.

 1. (**0.5 балла**) Определи минимальное уменьшение времени восстановления, которое можно будет статистически обнаружить с вероятностью $0.8$ при уровне значимости $\alpha=0.05$.

 2. (**0.5 балла**) На сколько оно изменится при изменении $\alpha$ с $5\%$ до $1\%$?


$$
H_0: \mu >= \mu_o \\
H_1: \mu < \mu_o \\
1 - \beta = 0.8 \\
\alpha \in \{0.05, 0.01\} 
$$

Проводим левосторонний тест, т.к "хотят проверить, уменьшает ли она среднее время восстановления после силовой нагрузки".

$$
MDE_{\alpha} = (Z_{1-\alpha} + Z_{1 - \beta}) \frac{\sigma}{\sqrt(n)} 
$$

```python
import scipy.stats as stats
from IPython.display import display, Math

n = 50
sigma = 36
z_095 = stats.norm.ppf(0.95)
z_099 = stats.norm.ppf(0.99)
z_08= stats.norm.ppf(0.8)

MDE_05 = (z_095 + z_08) * np.sqrt(sigma/n)
MDE_01 = (z_099 + z_08) * np.sqrt(sigma/n)

print(f"MDE_05 = {MDE_05:.3f}")
print(f"MDE_01 = {MDE_01:.3f}")
display(Math(rf'\Delta = {MDE_01 - MDE_05:.3f}'))
```

# (9 баллов) Задача 2. Новый лендинг продукта

В рамках программы по улучшению UX дизайна был значительно переработан лендинг нашего продукта.

 - Ранее конверсия из захода в покупку товара составляла $14.26\%$ (замер на $1250$ пользователей);

 - После переработки конверсия из захода в покупку товара составляет $17.9\%$ (замер на $890$ пользователей);

 - В финансовый план заложена цифра в $15\%$;

Можно ли утверждать, что:

 1. (**1.5 балл**) До изменения цифра не отклонялась статистически значимо от плана?
 2. (**1.5 балла**) До изменения цифра была меньше, чем цифра плана?
 3. (**1.5 балла**) После изменения цифра стала статистически значимо больше цифры плана?
 4. (**1.5 балла**) После изменения цифра не отклоняется статистически значимо от плана?
 5. (**1.5 балла**) После изменения цифра стала статистически значимо больше, чем была?

(**1.5 балла**) Для всех тестов также рассчитать мощность для различных вариантов обнаруживаемого эффекта.

```python
alpha = 0.05
```

```python
### Твоё решение здесь
p_before = 0.1426
n_before = 1250

p_after = 0.179
n_after = 890

p_plan = 0.15
```

### Пункт 1

$$H_0:p_{\text{до}}=0.15,\qquad H_1:p_{\text{до}}\ne0.15.$$

Для распределения Бернулли дисперсия равна $p(1-p)$. 

$$SE=\sqrt{\frac{\widehat p(1-\widehat p)}{n}},\qquad Z=\frac{\widehat p-p_0}{SE},$$

где $n=1250$, $\widehat p=0.1426$. а $p_0=0.15$ — план. 

Отвергаем $H_0$, если $|Z|>z_{\text{крит}}$.


```python
std_before = np.sqrt(p_before * (1 - p_before))
se_before = std_before / np.sqrt(n_before)
z_before = (p_before - p_plan) / se_before
z_critical = stats.norm.ppf(1 - alpha / 2)

print(f"Z = {z_before:.3f}")
print(f"Критические границы: {-z_critical:.3f}, {z_critical:.3f}")

if abs(z_before) > z_critical:
    print("Отвергаем H0: обнаружено статистически значимое отличие от плана.")
else:
    print("Не отвергаем H0: статистически значимого отличия от плана не обнаружено.")
```

**Вывод.** Получили $Z\approx-0.748$. Поскольку $|Z|=0.748<1.96$, статистика не попадает в критическую область, и нулевую гипотезу не отвергаем. Стат значимого отклонения конверсии от плана не обнаружено.


### Пункт 2

$$H_0:p_{\text{до}}\ge0.15,\qquad H_1:p_{\text{до}}<0.15.$$

Статистика та же, но тест левосторонний. Отвергаем $H_0$, если $Z<z_{\alpha}$.

```python
z_critical_left = stats.norm.ppf(alpha)

print(f"Z = {z_before:.3f}")
print(f"Критическая граница: {z_critical_left:.3f}")

if z_before < z_critical_left:
    print("Отвергаем H0: конверсия до изменения статистически значимо меньше плана.")
else:
    print("Не отвергаем H0: недостаточно оснований считать конверсию меньше плана.")
```

**Вывод.** Получили $Z\approx-0.748>-1.645$. Нулевую гипотезу не отвергаем. Хотя 14.26% меньше 15%, статистически значимого снижения относительно плана не обнаружено.

### Пункт 3

$$H_0:p_{\text{после}}\le0.15,\qquad H_1:p_{\text{после}}>0.15.$$

$$SE=\sqrt{\frac{\widehat p(1-\widehat p)}{n}},\qquad Z=\frac{\widehat p-p_0}{SE},$$

где $n=890$, $\widehat p=0.179$, $p_0=0.15$. Тест правосторонний. Отвергаем $H_0$, если $Z>z_{1-\alpha}$.

```python
std_after = np.sqrt(p_after * (1 - p_after))
se_after = std_after / np.sqrt(n_after)
z_after = (p_after - p_plan) / se_after
z_critical_right = stats.norm.ppf(1 - alpha)

print(f"Z = {z_after:.3f}")
print(f"Критическая граница: {z_critical_right:.3f}")

if z_after > z_critical_right:
    print("Отвергаем H0: конверсия после изменения статистически значимо больше плана.")
else:
    print("Не отвергаем H0: недостаточно оснований считать конверсию больше плана.")
```

**Вывод.** Получили $Z\approx2.257>1.645$. Нулевую гипотезу отвергаем. После изменения конверсия статистически значимо выше плана.

### Пункт 4

$$H_0:p_{\text{после}}=0.15,\qquad H_1:p_{\text{после}}\ne0.15.$$

Статистика та же, что в пункте 3, но тест двусторонний. Отвергаем $H_0$, если $|Z|>z_{1-\alpha/2}$.

```python
z_critical_two = stats.norm.ppf(1 - alpha / 2)

print(f"Z = {z_after:.3f}")
print(f"Критические границы: {-z_critical_two:.3f}, {z_critical_two:.3f}")

if abs(z_after) > z_critical_two:
    print("Отвергаем H0: конверсия после изменения статистически значимо отличается от плана.")
else:
    print("Не отвергаем H0: статистически значимого отличия от плана не обнаружено.")
```

**Вывод.** Получили $|Z|\approx2.257>1.96$. Нулевую гипотезу отвергаем. Утверждать, что значимого отклонения от плана нет, нельзя: отличие обнаружено.

### Пункт 5

$$H_0:p_{\text{после}}\le p_{\text{до}},\qquad H_1:p_{\text{после}}>p_{\text{до}}.$$

Считаем группы пользователей независимыми. Для двухвыборочного Z-теста:

$$SE=\sqrt{\frac{\widehat p_{\text{после}}(1-\widehat p_{\text{после}})}{n_{\text{после}}}+\frac{\widehat p_{\text{до}}(1-\widehat p_{\text{до}})}{n_{\text{до}}}},\qquad
Z=\frac{\widehat p_{\text{после}}-\widehat p_{\text{до}}}{SE}.$$

Тест правосторонний. Отвергаем $H_0$, если $Z>z_{1-\alpha}$.

```python
se_diff = np.sqrt(std_after**2 / n_after + std_before**2 / n_before)
z_diff = (p_after - p_before) / se_diff
z_critical_right = stats.norm.ppf(1 - alpha)

print(f"Z = {z_diff:.3f}")
print(f"Критическая граница: {z_critical_right:.3f}")

if z_diff > z_critical_right:
    print("Отвергаем H0: после изменения конверсия статистически значимо выше, чем до него.")
else:
    print("Не отвергаем H0: статистически значимого роста конверсии не обнаружено.")
```

**Вывод.** Получили $Z\approx2.245>1.645$. Нулевую гипотезу отвергаем. Конверсия выросла статистически значимо: с 14.26% до 17.9%, то есть на 3.64 процентного пункта.

Каждый пункт рассматриваем как отдельный тест при $\alpha=0.05$, без поправки на множественные проверки.

### Мощность тестов

```python
RANDOM_SEED = 42
```

```python
rng = np.random.default_rng(seed=RANDOM_SEED)
n_monte_carlo = 10000

def power_1sample(p0, p_true, n, alternative):
    p_sample = rng.binomial(n, p_true, size=n_monte_carlo) / n
    se = np.sqrt(p_sample * (1 - p_sample) / n)
    z = (p_sample - p0) / se

    if alternative == 'less':
        return np.mean(z < stats.norm.ppf(alpha))
    if alternative == 'greater':
        return np.mean(z > stats.norm.ppf(1 - alpha))
    return np.mean(np.abs(z) > stats.norm.ppf(1 - alpha / 2))

def power_2sample(p_before_true, p_after_true):
    before = rng.binomial(n_before, p_before_true, size=n_monte_carlo) / n_before
    after = rng.binomial(n_after, p_after_true, size=n_monte_carlo) / n_after
    se = np.sqrt(before * (1 - before) / n_before + after * (1 - after) / n_after)
    z = (after - before) / se
    return np.mean(z > stats.norm.ppf(1 - alpha))
```

```python
effects = [-0.05, -0.03, -0.02, -0.01, 0, 0.01, 0.02, 0.03, 0.05]
rows = []

for effect in effects:
    rows.append({
        'Эффект, п.п.': effect * 100,
        'Пункт 1': power_1sample(p_plan, p_plan + effect, n_before, 'two-sided'),
        'Пункт 2': power_1sample(p_plan, p_plan + effect, n_before, 'less'),
        'Пункт 3': power_1sample(p_plan, p_plan + effect, n_after, 'greater'),
        'Пункт 4': power_1sample(p_plan, p_plan + effect, n_after, 'two-sided'),
        'Пункт 5': power_2sample(p_before, p_before + effect),
    })

powers = pd.DataFrame(rows).set_index('Эффект, п.п.')
powers.round(3)
```
